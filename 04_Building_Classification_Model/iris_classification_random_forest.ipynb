## **1. Import Libraries**

```python
from sklearn import datasets #for iris dataset
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
import joblib
```

## **2. Load Iris data set**

```python
df = datasets.load_iris()
```

```python
df
```

## **3. Input Features**

The **iris** data set contains 4 input features and 1 output variable (the class label)

### **3.1. Input features**

```python
print(df.feature_names)
```

### **3.2. Output features**

```python
print(df.target_names)
```

```python
print(df.target)
```

## **4. Glimpse of Data**

### **4.1. Input Features**

```python
df.data
```

### **4.2. Output variable (the class label)**

```python
df.target
```

### **4.3. Assignin Input and Output variables**

```python
X = df.data
y = df.target
```

### **4.4. Let's examine the data dimension**

```python
X.shape
```

```python
y.shape
```

## **5. Build Classification Model Using Random Forest**

### Breakoown how can prediction work?

```python
model = RandomForestClassifier(
    bootstrap=True,
    ccp_alpha=0.0,
    class_weight=None,
    criterion='gini',
    max_depth=None,
    max_features='sqrt',
    max_leaf_nodes=None,
    max_samples=None,
    min_impurity_decrease=0.0,
    min_samples_leaf=1,
    min_samples_split=2,
    min_weight_fraction_leaf=0.0,
    n_estimators=100,
    n_jobs=None,
    oob_score=False,
    random_state=None,
    verbose=0,
    warm_start=False
)
```

warning: di baris model.fit(X, y) inilah tragedi Data Leakage (Kebocoran Kunci Jawaban) terjadi. Modelnya nggak belajar menebak, tapi murni menghafal 150 soal beserta jawabannya langsung (Overfitting parah)

```python
model.fit(X, y)
```

## **6. Feature Importance**

```python
print(model.feature_importances_)
```

## **7. Make Prediction**

```python
X
```

```python
y
```

```python
X.shape
```

```python
X[0], X[50], X[100]
```

```python
print(model.predict([[5.1, 3.5, 1.4, 0.2]]))
print(model.predict([[7. , 3.2, 4.7, 1.4]]))
print(model.predict([[6.3, 3.3, 6. , 2.5]]))
```

```python
y[0], y[50], y[100]
```

```python
print(model.predict_proba(X[[0, 50, 100]]))
```

## **8. Data Split (80/20 Ratio)**

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)
```

```python
X_train.shape, y_train.shape
```

```python
X_test.shape, y_test.shape
```

## **9. Rebuilt Random Forest Model**

```python
new_model = RandomForestClassifier()

new_model.fit(X_train, y_train)
```

### **9.1. Performs prediction samples**

```python
print(new_model.predict([[5.1, 3.5, 1.4, 0.2]]))
print(new_model.predict([[7. , 3.2, 4.7, 1.4]]))
print(new_model.predict([[6.3, 3.3, 6. , 2.5]]))
```

```python
print(new_model.predict_proba([[5.1, 3.5, 1.4, 0.2]]))
print(new_model.predict_proba([[7. , 3.2, 4.7, 1.4]]))
print(new_model.predict_proba([[6.3, 3.3, 6. , 2.5]]))
```

### **9.2. Performs prediction on the test set**

predicted class label

```python
print(new_model.predict(X_test))
```

```python
print(y_test)
```

```python
print(new_model.predict_proba(X_test))
```

## **10. Model Performance**

```python
print(new_model.score(X_test, y_test))
```

## **11. Save Model**

```python
joblib.dump(new_model, 'iris_classification_RF')
```
