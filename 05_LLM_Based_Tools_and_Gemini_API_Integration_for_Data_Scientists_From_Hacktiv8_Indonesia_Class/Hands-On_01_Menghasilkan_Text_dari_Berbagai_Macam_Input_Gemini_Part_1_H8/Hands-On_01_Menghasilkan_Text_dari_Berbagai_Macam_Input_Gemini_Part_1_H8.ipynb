# Document Understanding

Google Gemini juga dapat kita gunakan untuk mengelolah file atau dokumen dalam berbagai macam format, misalkan format pdf.

Untuk mempelajari lebih detail berkaitan dengan pengolahan Dokumen dengan Gemini, Anda dapat melihat pada referensi berikut:
- [Explore document processing capabilities with the Gemini API](https://ai.google.dev/gemini-api/docs/vision?lang=python)

````python
import httpx
import base64

model = genai.GenerativeModel("gemini-3.6-flash")
doc_url = "https://discovery.ucl.ac.uk/id/eprint/10089234/1/343019_3_art_0_py4t4l_convrt.pdf"  # Replace with the actual URL of your PDF

# Retrieve and encode the PDF
doc_data = base64.standard_b64encode(httpx.get(doc_url).content).decode("utf-8")

prompt = "Summarize this document"

response = model.generate_content([{'mime_type':'application/pdf', 'data': doc_data}, prompt])
print(response.text)
````

````python
from google import genai
from google.colab import userdata

# Pastikan API key sudah diambil (dari sel sebelumnya)
api_key = userdata.get('GOOGLE_API_KEY')

# Gunakan Client dari library SDK BARU!
client = genai.Client(api_key=api_key)

# Upload file dengan syntax SDK baru
print("Proses upload audio ke server Gemini...")
myfile = client.files.upload(file="sample.mp3")
print(f"✅ File berhasil diupload: {myfile.name}")

# Analisis pakai model yang direkomendasikan
print("Menganalisis audio, tunggu sebentar...")
response = client.models.generate_content(
    model="gemini-flash-latest", # Menggunakan model gemini-flash-latest
    contents=[myfile, "Describe this audio clip"]
)

print("\n--- HASIL ANALISIS AUDIO ---")
print(response.text)
````

# **Latihan**

Sekarang cobalah lakukan latihan untuk mengelolah gambar ataupun audio dengan data yang anda miliki (data dibebaskan). Serta lakukanlah beberapa macam teknik prompting dasar yang telah anda pelajari.

## **1. Role-Based-Prompting**
Teknik meminta model untuk mengadopsi dan menjawab dari perspektif peran atau profesi tertentu (misalnya bertindak sebagai ahli, dokter, atau ilmuwan) guna menyesuaikan gaya bahasa dan kedalaman analisisnya




````python
from google import genai
from google.colab import userdata
import PIL.Image

# 1. Panggil API Key dan pakai Client SDK BARU!
api_key = userdata.get('GOOGLE_API_KEY')
client = genai.Client(api_key=api_key)

# 2. Buka gambar (Pastikan path-nya udah bener di Colab lu)
image_path = "/content/335289b93881a34c2fc776c3ac0a334f.jpg"
img = PIL.Image.open(image_path)

# 3. Siapkan Prompt lu
role_based_prompt = "Sebagai seorang ahli cetologi, tolong jelaskan hewan apa yang ada di gambar dan jelaskan dengan cara yang mudah di pahami!"

# 4. Panggil model dan lempar gambar + teks secara bersamaan
print("Sedang menganalisis gambar paus, tunggu bentar Bos...")
response = client.models.generate_content(
    model="gemini-3.5-flash-lite",
    contents=[img, role_based_prompt]
)

print("\n--- HASIL ANALISIS GAMBAR ---")
print(response.text)
````

## **2. Priming (Contextual Prompting)**
Teknik memberikan latar belakang, cerita, skenario, atau informasi konteks awal yang relevan terlebih dahulu sebelum mengajukan pertanyaan atau perintah kepada model.

````python
priming_context = (
    "Konteks: Kapal penelitian biologi kelautan kita baru saja berlayar di Samudra Atlantik "
    "dan tim berhasil mengabadikan momen langka seekor mamalia laut raksasa yang sedang "
    "melakukan surface behavior di permukaan air.\n\n"
    "Berdasarkan skenario di atas, tolong identifikasi hewan yang ada di dalam gambar "
    "dan jelaskan dua perilaku unik utama yang sedang ditunjukkannya!"
)


print("Sedang memproses Priming Prompt dengan Gemini...")
response = client.models.generate_content(
    model="gemini-3.5-flash-lite",
    contents=[img, priming_context]
)

print("\n--- HASIL ANALISIS (PRIMING PROMPTING) ---")
print(response.text)
````

## **3. Comparative Prompting**
Teknik meminta model untuk membandingkan dua hal atau lebih secara side-by-side guna menjelaskan persamaan serta perbedaan di antara objek-objek tersebut.



````python
comparative_prompt = (
    "Berdasarkan gambar hewan yang sedang berenang di permukaan ini, "
    "lakukan **Comparative Prompting** dengan membandingkan antara Paus Bungkuk (hewan di gambar) "
    "dan Lumba-lumba Bottlenose dalam hal ukuran tubuh, struktur sirip dada, dan cara mereka berinteraksi atau berkomunikasi di lautan. "
    "Buat penjelasannya dalam bentuk poin-poin perbandingan yang jelas!"
)


print("Sedang memproses Comparative Prompt dengan Gemini...")
response = client.models.generate_content(
    model="gemini-3.5-flash-lite",
    contents=[img, comparative_prompt]
)

print("\n--- HASIL ANALISIS (COMPARATIVE PROMPTING) ---")
print(response.text)
````

## **4. Instructional Prompting**
Teknik memberikan instruksi yang jelas, terperinci, dan spesifik kepada model untuk melakukan tugas tertentu, sering kali disertai dengan urutan langkah-langkah kerja atau aturan format keluaran yang ketat.

````python

instructional_prompt = (
    "Tolong analisis gambar paus ini dengan mengikuti instruksi langkah-langkah ketat berikut:\n"
    "1. Identifikasi nama spesies secara ilmiah dan umum.\n"
    "2. Buat ringkasan ukuran tubuhnya dalam bentuk poin angka.\n"
    "3. Jelaskan secara spesifik fungsi biologis dari sirip dada yang terlihat pada gambar.\n"
    "4. Format seluruh jawaban menggunakan struktur Markdown dengan heading yang rapi."
)


print("Sedang memproses Instructional Prompt dengan Gemini...")
response = client.models.generate_content(
    model="gemini-3.5-flash-lite",
    contents=[img, instructional_prompt]
)

print("\n--- HASIL ANALISIS (INSTRUCTIONAL PROMPTING) ---")
print(response.text)
````

## **5. Few-Shot Prompting**
Teknik di mana model diberikan beberapa contoh (biasanya 2 sampai 5 contoh) pola atau format jawaban di awal untuk memandu model memahami ekspektasi sebelum mengerjakan tugas aslinya.

````python
few_shot_prompt = [
    "Contoh 1:\nInput: Foto lumba-lumba melompat.\nAnalisis: Hewan akuatik cerdas, mamalia, hidup berkelompok.\n",
    "Contoh 2:\nInput: Foto penyu di atas karang.\nAnalisis: Reptil purba bercangkang, bernapas dengan paru-paru, dilindungi.\n",
    "Sekarang giliran lu!\nInput: Foto paus bungkuk yang sedang berenang di permukaan air ini.\nTolong berikan analisis singkat dengan format yang persis sama seperti contoh di atas:"
]


print("Sedang memproses Few-Shot Prompt dengan Gemini...")
response = client.models.generate_content(
    model="gemini-3.5-flash-lite",
    contents=[img] + few_shot_prompt
)

print("\n--- HASIL ANALISIS (FEW-SHOT PROMPTING) ---")
print(response.text)
````

## **6. One-Shot Prompting**
Teknik di mana model diberikan tepat satu contoh kasus beserta format jawabannya terlebih dahulu sebelum diminta untuk melakukan tugas serupa pada input baru.

````python
one_shot_prompt = [
    "Contoh:\nInput: Foto lumba-lumba.\nKlasifikasi: Mamalia laut kecil, sangat sosial.\n\n",
    "Sekarang giliran lu!\nInput: Foto paus bungkuk di gambar ini.\nTolong berikan klasifikasi dengan format yang persis sama seperti contoh di atas:"
]


print("Sedang memproses One-Shot Prompt dengan Gemini...")
response = client.models.generate_content(
    model="gemini-3.5-flash-lite",
    contents=[img] + one_shot_prompt
)

print("\n--- HASIL ANALISIS (ONE-SHOT PROMPTING) ---")
print(response.text)
````

## **7. Zero-Shot Prompting**
Teknik di mana model diminta untuk menyelesaikan suatu tugas tanpa diberikan contoh sebelumnya atau penjelasan tambahan, murni hanya berdasarkan pertanyaan atau perintah langsung yang diajukan.

````python
zero_shot_prompt = "Jelaskan secara singkat jenis mamalia laut yang ada di gambar ini!"

print("Sedang memproses Zero-Shot Prompt dengan Gemini...")
response = client.models.generate_content(
    model="gemini-3.5-flash-lite",
    contents=[img, zero_shot_prompt]
)

print("\n--- HASIL ANALISIS (ZERO-SHOT PROMPTING) ---")
print(response.text)
````
