# **Hyperparametets for Machine Learning Model**

## **1. Make Syntetic Datasets**

### **1.1. Generate The Datasets**

```python
from sklearn.datasets import make_classification

X ,y = make_classification(n_samples=200, n_redundant=0, n_classes=2, n_features=10, random_state=42)
```

# **2. Data Split(80/20)**

### **2.1 Data Split**

ratio for testing = 20%
ration for training = 80%

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

### **2.2. Examine Data Dimension**

#### **2.2.1 Examine Data Train Dimension**

as we can see the training sets in X variable has 160 row with 10 column for the X while there are 160 row with 1 column in y variable

```python
X_train.shape, y_train.shape
```

#### **2.2.2 Examine Data Test Dimension**

the testing sets has 40 rows and 10 column in X variable and 40 rows in 1 column y yvariable

```python
X_test.shape, y_test.shape
```

## **3. Building a Simple Machine Learning Model with Random Forest**

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

rf = RandomForestClassifier(n_estimators=100, max_features=100)
```

```python
rf.fit(X_train, y_train)
```

```python
rf.score(X_test, y_test)
```

```python
y_pred = rf.predict(X_test)
```

```python
accuracy_score(y_pred, y_test)
```
